```python
ago = m[
    (m["Time_dt"] >= "2018-08-01") &
    (m["Time_dt"] < "2018-09-01")
].copy()

ago["delta"] = ago["Time_dt"].diff()

ago["delta"].value_counts().head(10)
```

```python
feb = m[
    (m["Time_dt"] >= "2019-02-01") &
    (m["Time_dt"] < "2019-03-01")
].copy()

feb["delta"] = feb["Time_dt"].diff()

feb["delta"].value_counts().head(10)
```

```python
periodo = m[
    (m["Time_dt"] >= "2018-08-01") &
    (m["Time_dt"] < "2019-01-01")
].copy()

periodo.shape
```

```python
numeric_cols = [
    "Indoor Temperature(°C)",
    "Outdoor Temperature(°C)",
    "Relative Pressure(hpa)",
    "Absolute Pressure(hpa)",
    "Wind Speed(km/h)",
    "Gust(km/h)",
    "DewPoint(°C)",
    "WindChill(°C)",
    "Hour Rainfall(mm)",
    "24 Hour Rainfall(mm)",
    "Week Rainfall(mm)",
    "Month Rainfall(mm)",
    "Total Rainfall(mm)",
    "Light(lux)"
]

for col in numeric_cols:
    m[col] = (
        m[col]
        .str.replace(".", "", regex=False)
        .str.replace(",", ".", regex=False)
    )
    m[col] = pd.to_numeric(m[col], errors="coerce")

for col in numeric_cols:
    m[col] = pd.to_numeric(m[col], errors="coerce")

missing = m.isna().sum().sort_values(ascending=False)

missing
```
